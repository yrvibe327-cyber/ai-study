## 01. 함수 정의

```python
######  1. 함수의  정의 유형 #########

#인자도 반환값도 없는 함수 정의
def my_func():    # 함수의 정의
    print("첫번째 함수입니다")
    print("인자도 반환값도 없는 함수 입니다")

my_func()         # 프로그램에서 사용하는 코드
```

```python
#인자는 있으나 반환값이 없는 함수
def my_friend(friendName):   # 함수의 정의
    print("{}는 나의 친구입니다.".format(friendName))

my_friend("철수")    # 프로그램에서 사용하는 코드
my_friend("영미")
```

```python
#인자도 있고, 반환값도 있는 함수
def my_calc(x, y):   #  함수의 정의
    z = x*y
    return z
a = my_calc(3,5)

print(a)
# or
print(my_calc(3,7))    #  프로그램 에서 사용 하는 코드
```

```python
######  2. 변수의 유효 범위 #########

a = 5  # 전역 변수

def func1():
    a = 1  # 지역 변수. func1()에서만 사용
    print("[func1] 지역 변수 a =", a)

def func2():
    a = 2  # 지역 변수. func2()에서만 사용
    print("[func2] 지역 변수 a =", a)

def func3():
    print("[func3] 전역 변수 a =", a)

def func4():
    global a  # 함수 내에서 전역 변수 변경 위해 선언
    a = 4  # 전역 변수의 값 변경
    print("[func4] 전역 변수 a =", a)

func1() #함수 func1() 호출
func2() #함수 func2() 호출
func3() #함수 func3() 호출
func4() #함수 func4() 호출
```

```python
######  3. 람다 함수 #########
(lambda x : x**2) (3)
print("람다 함수 수행결과는 {0}입니다".format((lambda x : x**2) (3)))

mySquare = lambda x : x**2
mySquare(2)
print("람다 함수 수행결과는 {0}입니다".format(mySquare(2)))

mySimpleFunc = lambda x, y, z: 2 * x + 3 * y + z
# mySimpleFunc(1, 2, 3)
print("람다 함수 수행결과는 {0}입니다".format(mySimpleFunc(1, 2, 3)))
```

```python
##########################################################################################
# Print 함수 : print("문자열", sep 옵션, end 옵션)
#             sep(seperator) 옵션 : 출력시 출력값들 사이에 넣어줄 구분자(기본값이 공백)
#             end 옵션 : 마지막 문자를 출력하고 그다음에 출력할 문자(기본값이 줄을바꾸는 개행 특수문자이다)

# print sep 옵션 변경 예제
print('aa', 'bb', 'cc')         # 기본 값대로 출력
print('-----------------------------------')
print('aa', 'bb', 'cc', sep='_') # 출력값들사이의 구분자를 공백에서 "_" 로 바꾸었다.
```

```python
# print end 옵션 변경 예제
print('aa')                      # 기본 값대로 출력
print('bb')
print('cc')
print('-----------------------------------')
print('aa', end='\n\n') # 개행문자 출력이 하나이던 기본값을 3개의 개행문자 출력으로 바꾸었다.
print('bb', end='\n\n\n') # 개행문자 출력이 하나이던 기본값을 3개의 개행문자 출력으로 바꾸었다.
print('cc')               # 개행문자 출력이 하나이던 기본값을 3개의 개행문자 출력으로 바꾸었다.
```

```python

#############내장함수들을 이용한 간단한 활용프로그램 만들어보기 ###########

## < 내장함수 를 사용하지 않은 경우 프로그램 >
scores = [90, 80, 95, 85]  # 과목별 시험 점수

score_sum = 0  # 총점 계산을 위한 초깃값 설정
subject_num = 0  # 과목수 계산을 위한 초깃값 설정
for score in scores:
    score_sum = score_sum + score  # 과목별 점수 모두 더하기
    subject_num = subject_num + 1  # 과목수 계산

average = score_sum / subject_num  # 평균(총점 / 과목수) 구하기

print("총점:{0}, 평균:{1}".format(score_sum, average))

## < 내장함수 를 사용한 경우 프로그램 > ==> 상당히 프로그램 수가 줄어든다.

scores = [90, 80, 95, 85] # 과목별 시험 점수

print("총점:{0}, 평균:{1}".format(sum(scores), sum(scores)/len(scores)))
```
