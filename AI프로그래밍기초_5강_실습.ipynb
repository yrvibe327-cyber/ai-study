## 01. Class 정의

```python
## 클래스 정의
class Bicycle():

    wheel_size = 26  # 객체의 속성 설정
    color = 'black'  # 객체의 속성 설정

    def move(self, speed):
        print("자전거: 시속 {0}킬로미터로 전진".format(speed))

    def turn(self, direction):
        print("자전거: {0}회전".format(direction))

    def stop(self):
        print("자전거({0}, {1}): 정지 ".format(self.wheel_size, self.color))
```

```python
## 객체 생성

my_bicycle1 = Bicycle()     # Bicycle 클래스의 인스턴스인 my_bicycle 객체 생성
print("첫번째 로 생성한 객체 입니다.")
my_bicycle1.move(30)  # 객체의 메서드 호출
my_bicycle1.turn('좌')
my_bicycle1.stop()
```

```python
## 두번째 객체 생성

my_bicycle2 = Bicycle() # Bicycle 클래스의 인스턴스인 my_bicycle 객체 생성

print("--------------------------")
print("두번째로 생성한 객체 입니다.")

my_bicycle2.wheel_size = 35 # 객체의 속성 설정
my_bicycle2.color = 'blue'

my_bicycle2.move(30)  # 객체의 메서드 호출
my_bicycle2.turn('좌')
my_bicycle2.stop()
```

## 02. Class 변수, instance 변수

```python
class Car():
    count = 0;  # 클래스 변수 생성 및 초기화

    def __init__(self, size, color):
        self.size = size    # 인스턴스 변수 생성 및 초기화
        self.color = color  # 인스턴스 변수 생성 및 초기화
        self.count = 0      # 인스턴스 변수 생성 및 초기화
        Car.count = Car.count + 1  # 클래스 변수 이용
        print("자동차 객체의 수: Car.count = {0}".format(Car.count))
        print("인스턴스 변수 초기화: self.count = {0}".format(self.count))

    def move(self):
        print("자동차({0} & {1})가 움직입니다.".format(self.size, self.color))
```

```python
car1 = Car("big", 'white')
car1.move()
```

```python
car2 = Car("small", 'black')
car2.move()
```

## 03. 정적 메서드

```python
# Car 클래스 선언
class Car():

    # 정적 메서드
    @staticmethod
    def check_type(model_code):
        if (model_code >= 20):
            print("이 자동차는 전기차입니다.")
        elif (10 <= model_code < 20):
            print("이 자동차는 가솔린차입니다.")
        else:
            print("이 자동차는 디젤차입니다.")


Car.check_type(25)
Car.check_type(2)
```

## 04. Class 메서드

```python
# Car 클래스 선언
class Car():
    instance_count = 0  # 클래스 변수

    # 초기화 함수(인스턴스 메서드)
    def __init__(self, size, color):
        self.size = size  # 인스턴스 변수
        self.color = color  # 인스턴스 변수
        Car.instance_count = Car.instance_count + 1


    # 클래스 메서드
    @classmethod
    def count_instance(cls):    # "cls" 인자의 의미:  Class 자체를 인자로 받는다
        print("자동차 객체의 개수: {0}".format(cls.instance_count))


Car.count_instance()        # 객체 생성 전에 클래스 메서드 호출(중요)
Car.count_instance()        # 객체 생성 전에 클래스 메서드 호출(중요)
Car.count_instance()        # 객체 생성 전에 클래스 메서드 호출(중요)

car1 = Car("small", "red")  # 첫 번째 객체 생성
Car.count_instance()        # 클래스 메서드 호출

car2 = Car("big", "green")  # 두 번째 객체 생성
Car.count_instance()        # 클래스 메서드 호출
```

## 05. Class를 사용하지 않는 프로그램

```python
# class를 사용하지 않는 프로그램과 class를 사용하는 프로그램과의 비교
# 1. class를 사용하지 않는 프로그램 예제

robot1_name = 'R1' # 로봇 이름
robot1_pos = 0     # 로봇의 초기 위치

def robot1_move():
    global robot1_pos
    robot1_pos = robot1_pos + 1
    print("{0} position: {1}".format(robot1_name, robot1_pos))

robot2_name = 'R2' # 로봇 이름
robot2_pos = 10 # 로봇의 초기 위치

def robot2_move():
    global robot2_pos
    robot2_pos = robot2_pos + 1
    print("{0} position: {1}".format(robot2_name, robot2_pos))

robot1_move()
robot2_move()
# Robot이 두대로 늘어남에 따라 변수와 함수 코딩량이 두배로 늘어났음,
# 만약 Robot이 10대 로 늘어난다면 변수는 20개 함수는 10개 를 코딩해야함.
```

```python
# class를 사용하지 않는 프로그램과 class를 사용하는 프로그램과의 비교
# 2. class를 사용하는 프로그램 예제

class Robot():
    def __init__(self, name, pos):
        self.name = name  # 로봇 객체의 이름
        self.pos = pos  # 로봇 객체의 위치

    def move(self):
        self.pos = self.pos + 1
        print("{0} position: {1}".format(self.name, self.pos))

robot1 = Robot('R1', 0)
robot2 = Robot('R2', 10)
robot3 = Robot('R3', 15)
robot4 = Robot('R2', 20)
robot1.move()
robot2.move()
robot3.move()
robot4.move()
# Robot을 여러개 만들어도 함수 정의에서 변수나 함수의 수가 증가 하지 않는다.
```

## 07. Class 상속

```python
# 부모 클래스 (기 정의된)
class Bicycle():

    def __init__(self, wheel_size, color):
        self.wheel_size = wheel_size
        self.color = color

    def move(self, speed):
        print("자전거: 시속 {0}킬로미터로 전진".format(speed))

    def turn(self, direction):
        print("자전거: {0}회전".format(direction))

    def stop(self):
        print("자전거({0}, {1}): 정지 ".format(self.wheel_size, self.color))

# 자식 클래스 정의
class FoldingBicycle(Bicycle): # 부모 Class 상속

    def __init__(self, wheel_size, color, state):  # FoldingBicycle 초기화
        Bicycle.__init__(self, wheel_size, color)  # Bicycle의 초기화 재사용
        # super().__init__(wheel_size, color) # super()도 사용 가능
        self.state = state  # 자식 클래스에서 새로 추가한 변수

    def fold(self):
        self.state = 'folding'
        print("자전거: 접기, state = {0}".format(self.state))

    def unfold(self):
        self.state = 'unfolding'
        print("자전거: 펴기, state = {0}".format(self.state))

```

```python
# 자식클라스 이용 객체 생성
folding_bicycle = FoldingBicycle(27, 'white', 'unfolding') # 객체 생성

folding_bicycle.move(20) # 부모 클래스의 함수(메서드) 호출
folding_bicycle.fold()   # 자식 클래스에서 정의한 함수 호출
folding_bicycle.unfold()
```
